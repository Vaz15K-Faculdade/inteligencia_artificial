##**Prática de Aprendizado Supervisionado: classificação, regressão e avaliação de modelos**

Objetivos da atividade

- diferenciar problemas de classificação e regressão;
- separar corretamente os dados em treino e teste;
- treinar modelos supervisionados usando bibliotecas;
- avaliar o desempenho com métricas apropriadas;
- interpretar os resultados e justificar qual modelo foi mais adequado.
### Bibliotecas
* pandas
* numpy
* matplotlib
* scikit-learn

## Parte 1 — Classificação

Usar o dataset Breast Cancer Wisconsin do scikit-learn.

Modelos:

* Regressão Logística
* k-NN

Métricas:

* acurácia
* precisão
* recall
* F1-score
* matriz de confusão

## Parte 2 — Regressão

Usar o dataset Diabetes do scikit-learn.

Modelos:

* Regressão Linear
* Árvore de Regressão

Métricas:

* MAE
* MSE
* RMSE
* R²

## Parte 3 — Interpretação

- qual problema era de classificação e qual era de regressão;

- por que a métrica acurácia não serve para regressão;

- por que RMSE não é a métrica principal para classificação;

- qual modelo teve melhor desempenho em cada tarefa;

- se houve indício de sobreajuste ao comparar treino e teste.

```python
# ============================================================
# ATIVIDADE PRÁTICA - APRENDIZADO SUPERVISIONADO
# Classificação e Regressão com separação de dados
# ============================================================

# =========================
# 1. IMPORTAÇÃO DE BIBLIOTECAS
# =========================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_breast_cancer, load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Modelos de classificação
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier

# Modelos de regressão
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor

# Métricas de classificação
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report
)

# Métricas de regressão
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Reprodutibilidade
RANDOM_STATE = 42


# ============================================================
# PARTE A - CLASSIFICAÇÃO
# Objetivo: prever se o tumor é maligno ou benigno
# ============================================================

print("=" * 70)
print("PARTE A - CLASSIFICAÇÃO")
print("=" * 70)

# -------------------------
# 2. CARREGAR O DATASET DE CLASSIFICAÇÃO
# -------------------------
cancer = load_breast_cancer()

# Criar DataFrame para facilitar visualização
df_cancer = pd.DataFrame(cancer.data, columns=cancer.feature_names)
df_cancer["target"] = cancer.target

print("\nPrimeiras linhas do dataset de classificação:")
display(df_cancer.head())

print("\nInformações gerais:")
print(df_cancer.info())

print("\nDistribuição das classes:")
print(df_cancer["target"].value_counts())

# target = 0 (maligno), 1 (benigno)
print("\nNome das classes:", cancer.target_names)

# -------------------------
# 3. DEFINIR X E y
# -------------------------
X_class = cancer.data
y_class = cancer.target

# -------------------------
# 4. SEPARAÇÃO TREINO/TESTE
# stratify=y_class preserva a proporção das classes
# -------------------------
Xc_train, Xc_test, yc_train, yc_test = train_test_split(
    X_class, y_class,
    test_size=0.2,
    random_state=RANDOM_STATE,
    stratify=y_class
)

print(f"\nTamanho treino: {Xc_train.shape[0]} amostras")
print(f"Tamanho teste : {Xc_test.shape[0]} amostras")

# -------------------------
# 5. PADRONIZAÇÃO DOS DADOS
# Importante para modelos baseados em distância e escala
# -------------------------
scaler_class = StandardScaler()

Xc_train_scaled = scaler_class.fit_transform(Xc_train)
Xc_test_scaled = scaler_class.transform(Xc_test)

# -------------------------
# 6. TREINAR MODELOS DE CLASSIFICAÇÃO
# -------------------------
modelo_log = LogisticRegression(max_iter=10000, random_state=RANDOM_STATE)
modelo_knn = KNeighborsClassifier(n_neighbors=5)

modelo_log.fit(Xc_train_scaled, yc_train)
modelo_knn.fit(Xc_train_scaled, yc_train)

# -------------------------
# 7. PREDIÇÕES
# -------------------------
yc_pred_log = modelo_log.predict(Xc_test_scaled)
yc_pred_knn = modelo_knn.predict(Xc_test_scaled)

# Também vamos calcular desempenho no treino para observar possível overfitting
yc_train_pred_log = modelo_log.predict(Xc_train_scaled)
yc_train_pred_knn = modelo_knn.predict(Xc_train_scaled)

# -------------------------
# 8. FUNÇÃO PARA AVALIAÇÃO DE CLASSIFICAÇÃO
# -------------------------
def avaliar_classificacao(nome_modelo, y_train, y_train_pred, y_test, y_test_pred):
    print("\n" + "-" * 60)
    print(f"Modelo: {nome_modelo}")
    print("-" * 60)

    acc_train = accuracy_score(y_train, y_train_pred)
    acc_test = accuracy_score(y_test, y_test_pred)
    prec = precision_score(y_test, y_test_pred)
    rec = recall_score(y_test, y_test_pred)
    f1 = f1_score(y_test, y_test_pred)

    print(f"Acurácia treino: {acc_train:.4f}")
    print(f"Acurácia teste : {acc_test:.4f}")
    print(f"Precisão       : {prec:.4f}")
    print(f"Recall         : {rec:.4f}")
    print(f"F1-score       : {f1:.4f}")

    print("\nRelatório de classificação:")
    print(classification_report(y_test, y_test_pred, target_names=cancer.target_names))

    cm = confusion_matrix(y_test, y_test_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=cancer.target_names)
    disp.plot()
    plt.title(f"Matriz de Confusão - {nome_modelo}")
    plt.show()

# -------------------------
# 9. AVALIAR MODELOS DE CLASSIFICAÇÃO
# -------------------------
avaliar_classificacao("Regressão Logística", yc_train, yc_train_pred_log, yc_test, yc_pred_log)
avaliar_classificacao("k-NN", yc_train, yc_train_pred_knn, yc_test, yc_pred_knn)


# ============================================================
# PARTE B - REGRESSÃO
# Objetivo: prever progressão quantitativa da doença
# ============================================================

print("=" * 70)
print("PARTE B - REGRESSÃO")
print("=" * 70)

# -------------------------
# 10. CARREGAR O DATASET DE REGRESSÃO
# -------------------------
diabetes = load_diabetes()

df_diabetes = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)
df_diabetes["target"] = diabetes.target

print("\nPrimeiras linhas do dataset de regressão:")
display(df_diabetes.head())

print("\nInformações gerais:")
print(df_diabetes.info())

print("\nDescrição estatística:")
display(df_diabetes.describe())

# -------------------------
# 11. DEFINIR X E y
# -------------------------
X_reg = diabetes.data
y_reg = diabetes.target

# -------------------------
# 12. SEPARAÇÃO TREINO/TESTE
# -------------------------
Xr_train, Xr_test, yr_train, yr_test = train_test_split(
    X_reg, y_reg,
    test_size=0.2,
    random_state=RANDOM_STATE
)

print(f"\nTamanho treino: {Xr_train.shape[0]} amostras")
print(f"Tamanho teste : {Xr_test.shape[0]} amostras")

# -------------------------
# 13. PADRONIZAÇÃO
# Para regressão linear é útil; para árvore não é obrigatória,
# mas usaremos os mesmos dados escalados na regressão linear.
# -------------------------
scaler_reg = StandardScaler()

Xr_train_scaled = scaler_reg.fit_transform(Xr_train)
Xr_test_scaled = scaler_reg.transform(Xr_test)

# -------------------------
# 14. TREINAR MODELOS DE REGRESSÃO
# -------------------------
modelo_lr = LinearRegression()
modelo_tree = DecisionTreeRegressor(max_depth=4, random_state=RANDOM_STATE)

modelo_lr.fit(Xr_train_scaled, yr_train)
modelo_tree.fit(Xr_train, yr_train)

# -------------------------
# 15. PREDIÇÕES
# -------------------------
yr_pred_lr = modelo_lr.predict(Xr_test_scaled)
yr_pred_tree = modelo_tree.predict(Xr_test)

yr_train_pred_lr = modelo_lr.predict(Xr_train_scaled)
yr_train_pred_tree = modelo_tree.predict(Xr_train)

# -------------------------
# 16. FUNÇÃO PARA AVALIAÇÃO DE REGRESSÃO
# -------------------------
def avaliar_regressao(nome_modelo, y_train, y_train_pred, y_test, y_test_pred):
    print("\n" + "-" * 60)
    print(f"Modelo: {nome_modelo}")
    print("-" * 60)

    mae_train = mean_absolute_error(y_train, y_train_pred)
    mse_train = mean_squared_error(y_train, y_train_pred)
    rmse_train = np.sqrt(mse_train)
    r2_train = r2_score(y_train, y_train_pred)

    mae_test = mean_absolute_error(y_test, y_test_pred)
    mse_test = mean_squared_error(y_test, y_test_pred)
    rmse_test = np.sqrt(mse_test)
    r2_test = r2_score(y_test, y_test_pred)

    print("Treino:")
    print(f"MAE  : {mae_train:.4f}")
    print(f"MSE  : {mse_train:.4f}")
    print(f"RMSE : {rmse_train:.4f}")
    print(f"R²   : {r2_train:.4f}")

    print("\nTeste:")
    print(f"MAE  : {mae_test:.4f}")
    print(f"MSE  : {mse_test:.4f}")
    print(f"RMSE : {rmse_test:.4f}")
    print(f"R²   : {r2_test:.4f}")

    # Gráfico valor real x valor predito
    plt.figure(figsize=(7,5))
    plt.scatter(y_test, y_test_pred)
    plt.xlabel("Valor real")
    plt.ylabel("Valor predito")
    plt.title(f"Valores Reais x Preditos - {nome_modelo}")
    plt.grid(True)
    plt.show()

# -------------------------
# 17. AVALIAR MODELOS DE REGRESSÃO
# -------------------------
avaliar_regressao("Regressão Linear", yr_train, yr_train_pred_lr, yr_test, yr_pred_lr)
avaliar_regressao("Árvore de Regressão", yr_train, yr_train_pred_tree, yr_test, yr_pred_tree)

# -------------------------
# 18. COMPARAÇÃO FINAL DOS MODELOS
# -------------------------
resultados = pd.DataFrame({
    "Tarefa": ["Classificação", "Classificação", "Regressão", "Regressão"],
    "Modelo": ["Regressão Logística", "k-NN", "Regressão Linear", "Árvore de Regressão"],
    "Métrica principal": [
        accuracy_score(yc_test, yc_pred_log),
        accuracy_score(yc_test, yc_pred_knn),
        np.sqrt(mean_squared_error(yr_test, yr_pred_lr)),
        np.sqrt(mean_squared_error(yr_test, yr_pred_tree))
    ]
})

print("\nResumo dos resultados:")
display(resultados)

# ============================================================
# 19. INTERPRETAÇÃO AUTOMÁTICA BÁSICA
# ============================================================
print("=" * 70)
print("INTERPRETAÇÃO")
print("=" * 70)

print("""
1. Em CLASSIFICAÇÃO, a saída é uma classe/categoria.
   Exemplo: benigno ou maligno.
   Por isso usamos acurácia, precisão, recall e F1-score.

2. Em REGRESSÃO, a saída é um valor numérico contínuo.
   Exemplo: previsão de um valor.
   Por isso usamos MAE, MSE, RMSE e R².

3. A escolha do modelo depende do problema:
   - Se a variável-alvo é categórica -> classificação
   - Se a variável-alvo é numérica contínua -> regressão

4. Comparar desempenho em treino e teste ajuda a perceber
   se o modelo generaliza bem ou se está ajustado demais ao treino.
""")
```
