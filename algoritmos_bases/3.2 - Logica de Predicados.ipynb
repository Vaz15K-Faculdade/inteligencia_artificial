###**Conteúdo Teórico**
#### 1. Revisão rápida: Lógica Proposicional
* Proposição: enunciado verdadeiro ou falso.
* Conectivos lógicos básicos: ¬, ∧, ∨, →, ↔.
* Inferência Modus Ponens (P→Q, P => Q).

#### 2. Por que lógica de predicados?
* Vai além da proposicional, pois permite raciocinar sobre objetos, propriedades e relações.
* Usa quantificadores:
  * Universal (∀): "Para todo x"
  * Existencial (∃): "Existe um x"
* Permite formalizar afirmações complexas, exemplo:
  ∀x(Humano(x)→Mortal(x))
  "Todos os humanos são mortais".

#### 3. Elementos da lógica de predicados
* Variáveis, constantes, predicados, funções.
* Termos: constantes, variáveis, ou funções aplicadas a termos.
* Sentenças atômicas e compostas.
* Quantificadores e escopo.

####. 4. Semântica
* Sentenças são avaliadas no contexto de domínios e interpretações.
* Verdadeiro ou falso depende da atribuição dos valores aos termos no domínio.

#### Prática em Python: Inferência Simples por Encadeamento
1. Simulação básica de inferência (forward chaining)




```python
# Definição simples de regras e fatos
regras = {
    "humano": "mortal",
    "ave": "bípede",
    "peixe": "aquático"
}

fatos = ["socrates é humano", "platao é humano", "nemo é peixe"]

# Encadeamento simples: gerar conclusões a partir das regras e fatos
for fato in fatos:
    for pred, concl in regras.items():
        if pred in fato:
            sujeito = fato.split()[0]
            print(f"{sujeito} é {concl}")
```

### Modelagem de conhecimento com lógica de predicados


```python
# Definição de fatos e regras expressos simbolicamente
populacao = {
    'pessoas': ['Alice', 'Bob', 'Carol', 'Dave'],
    'animais': ['Fido', 'Toby'],
}

# Predicados (funções que definimos):
# humano(x), cachorro(y), gosta(x,y)
# Função para armazenar fatos:
humano = {'Alice', 'Bob', 'Carol'}
cachorro = {'Fido', 'Toby'}

# Relação "gosta" entre pessoa e animal
gosta = {
    ('Alice', 'Fido'),
    ('Bob', 'Toby'),
    ('Carol', 'Fido'),
    ('Dave', 'Toby')  # Dave is not declared humano? To check inference.
}

print("População humana:", humano)
print("População canina:", cachorro)
print("Relações gosta:", gosta)
```

### Inferência simples: quem gosta de cachorro e é humano?

```python
# Inferência: Para todo x, se humano(x) e gosta(x,y) e cachorro(y), então "x gosta de cachorro" válida.

for pessoa, animal in gosta:
    if pessoa in humano and animal in cachorro:
        print(f"{pessoa} é humano e gosta do cachorro {animal}.")
```

### Exercício: adicione fatos e veja inferência


```python
# Pergunta: Dave é humano? O que ocorre com a inferência?
# Adicione Dave aos humanos e veja a mudança

humano.add('Dave')

for pessoa, animal in gosta:
    if pessoa in humano and animal in cachorro:
        print(f"{pessoa} é humano e gosta do cachorro {animal}.")
```

### Simples inferência em regras codificadas (forward chaining)

```python
# Base de regras: se gosta de cachorro e humano, se feliz
regras = {
    'Se gosta de cachorro e é humano': 'é feliz'
}

# Inferir felicidade das pessoas:
felizes = set()

for pessoa, animal in gosta:
    if pessoa in humano and animal in cachorro:
        felizes.add(pessoa)

for pessoa in felizes:
    print(f"{pessoa} {regras['Se gosta de cachorro e é humano']}")
```

### Modelagem com quantificadores utilizando Python (interatividade)

```python
# Verifique se "Todos os humanos gostam de pelo menos um cachorro"

todos_gostam = True
for pessoa in humano:
    gosta_de_algum = any((pessoa, animal) in gosta for animal in cachorro)
    if not gosta_de_algum:
        todos_gostam = False
        print(f"{pessoa} não gosta de nenhum cachorro.")
        break

if todos_gostam:
    print("Todos os humanos gostam de pelo menos um cachorro.")
```

### Extensão: integração com Sympy para lógica proposicional

```python
from sympy import symbols
from sympy.logic import boolalg, satisfiable

P, Q = symbols('P Q')

# Exemplo diferente: Uma regra complexa (P ∧ Q) → ¬P
expr = boolalg.Implies(boolalg.And(P,Q), ~P)

print("Expressão:", expr)
print("Tabela verdade para expr:")
print(list(satisfiable(expr, all_models=True)))
```
