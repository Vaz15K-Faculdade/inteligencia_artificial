```python
import math
import heapq
import networkx as nx
```

```python
def construir_grid_grafo(linhas, colunas, obstaculos=set()):
  G = nx.grid_2d_graph(linhas, colunas) # 4 vizinhos
  # remove obstaculos (nos)
  G.remove_nodes_from(obstaculos)
  # torna ponderado com custo 1 por aresta
  for u, v in G.edges():
    G[u][v]["weight"] = 1
  return G

def manhattan(a,b): # h admissivel para 4-direções com custo 1
  return abs(a[0]-b[0] + abs(a[1]-b[1]))
```

```python
def greedy_best_first(G, start, goal, h):
  pq = []
  heapq.heappush(pq, (h(start, goal), start))
  came = {start: None}
  visit = {start}
  exp = 0

  while pq:
    _, u = heapq.heappop(pq); exp += 1
    if u == goal:
      # reconstroi caminho
      path = []

      while u is not None:
        path.append(u); u = came[u]
      return list(reversed(path)), exp

      for v in G.neighborns(u):
        if v not in visit:
          visit.add(v)
          came[v] = u
          heapq.heappush(pq, (h(v, goal), v))
  return None, exp
```

```python
linhas, colunas = 6, 8
start, goal = (0,0), (5,7)
obst = {(0,5), (1,1), (1,2), (1,5), (2,7), (3,0), (3,1), (3,3), (4,3), (4,5), (5,1)}

GG = construir_grid_grafo(linhas, colunas, obstaculos=obst)
```

```python
caminho_astar = nx.astar_path(GG, start, goal, heuristic=manhattan, weight="weight")

print(f"A* (manhattan) caminho: {caminho_astar}")
print(f"A* custo: {len(caminho_astar)-1}")
```

```python
caminho_greedy, exp_greedy = greedy_best_first(GG, start, goal, h=manhattan)

print(f"Gulosa caminho: {caminho_greedy}")
print(f"Gulosa passadas (expansões): {exp_greedy}")
```
