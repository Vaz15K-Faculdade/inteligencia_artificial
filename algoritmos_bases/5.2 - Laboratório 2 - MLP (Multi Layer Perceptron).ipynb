# Laboratório 2: MLP (Multi-Layer Perceptron)


*Objetivo*: Classificação de dados tabulares (exemplo: Iris).
- Construir redes com múltiplas camadas ocultas.
- Usar funções de ativação ReLU e saída sigmoid/softmax.
- Aplicar normalização de entrada e validação cruzada.

*Pratica*
- Testar arquiteturas diferentes (nº camadas, neurônios).
- Usar callback EarlyStopping para evitar overfitting.
- Analisar métricas acurácia e perda.

O MLP possui múltiplas camadas ocultas e funções de ativação não lineares, permitindo aprender padrões complexos.
Parâmetros importantes: número de camadas, neurônios por camada, função de ativação, regularização.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.layers import Dense
from tensorflow.keras.models import Sequential

# Carregando conjunto Iris
X, y = load_iris(return_X_y=True)
y_cat = to_categorical(y)  # codificação one-hot para saída multi-classe

# Dividindo em treino/validação (80%-20%)
X_train, X_test, y_train, y_test = train_test_split(X, y_cat, test_size=0.2, random_state=42)

# Construindo MLP
model = Sequential([
    Dense(16, activation='relu', input_shape=(X.shape[1],)),  # camada oculta 1: 16 neurônios, ReLU
    Dense(16, activation='relu'),  # camada oculta 2: igual
    Dense(3, activation='softmax') # camada saída: 3 neurônios (classes), softmax para distribuição
])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Treinamento com validação
model.fit(X_train, y_train, epochs=50, validation_split=0.1, verbose=1)

# Avaliação em teste
loss, acc = model.evaluate(X_test, y_test)
print(f"Acurácia: {acc:.2f}")

# Comentários:
# - ReLU ativa só valores positivos e ajuda evitar 'desaparecimento do gradiente'
# - Softmax transforma saída em distribuição probalística (soma = 1)
# - Adam combina vantagens do SGD adaptando taxa de aprendizado para cada peso
# - Camadas ocultas permitem modelar padrões complexos e não lineares
```

- Teste diferentes números de camadas e neurônios.
- Use EarlyStopping para evitar overfitting.

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

# Dados XOR
entrada = np.array([[0,0], [0,1], [1,0], [1,1]])
saida_esperada = np.array([[0], [1], [1], [0]])

# Construção da MLP simples com Input explicitamente definido
modelo = Sequential([
    Input(shape=(2,)),                # Especifica o formato da entrada
    Dense(2, activation='sigmoid'),   # Camada oculta com 2 neurônios
    Dense(1, activation='sigmoid')    # Saída com 1 neurônio
])

# Compilar modelo com otimizador SGD e perda binária
modelo.compile(optimizer='sgd', loss='binary_crossentropy', metrics=['accuracy'])

# Treinamento
modelo.fit(entrada, saida_esperada, epochs=10000, verbose=0)

# Avaliação
perda, acuracia = modelo.evaluate(entrada, saida_esperada, verbose=0)
print(f"Acurácia no problema XOR: {acuracia*100:.2f}%")

# Previsões
previsoes = modelo.predict(entrada)
print("Previsões para entradas XOR:")
for i, p in enumerate(previsoes):
    print(f"Entrada: {entrada[i]} => Saída prevista (probabilidade): {p[0]:.4f}")
```

- Aumentar o número de neurônios na camada oculta (ex: 4 ou 8).
- Usar ReLU ou tanh em vez de sigmoid na oculta.
- Aumentar as épocas ou ajustar a taxa de aprendizado.
