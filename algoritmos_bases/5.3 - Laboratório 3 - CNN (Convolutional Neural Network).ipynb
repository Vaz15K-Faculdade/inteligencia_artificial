#Laboratório 3: CNN (Convolutional Neural Network)

*Objetivo*: Classificação de imagens (exemplo: MNIST).

**Pratica**
- Ajustar hiperparâmetros: filtros, tamanho kernal, pooling.
- Experimentar data augmentation básico.
- Avaliar precisão e histórico de perda.

A CNN utiliza camadas convolucionais para extrair características locais das imagens.
Parâmetros importantes: número de filtros, tamanho do kernel, pooling, camadas densas.


```python
from tensorflow.keras.datasets import mnist
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from tensorflow.keras.layers import Dense
from tensorflow.keras.models import Sequential

# Carregando MNIST e preprocessamento
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()
train_images = train_images.reshape(-1, 28, 28, 1).astype('float32') / 255
test_images = test_images.reshape(-1, 28, 28, 1).astype('float32') / 255

# Criando a rede CNN
model = Sequential([
    # Camada convolucional com 32 filtros 3x3 e ReLU, entrada 28x28x1
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),

    # MaxPooling reduz dimensionalidade pela metade, ajuda o modelo a generalizar
    MaxPooling2D(2),

    # Flatten converte volume 2D em vetor para camadas densas
    Flatten(),

    # Camada densa intermediária
    Dense(64, activation='relu'),

    # Camada de saída para 10 classes, softmax para probabilidade
    Dense(10, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Treinamento, usando 10% para validação e 5 épocas (pode aumentar)
model.fit(train_images, train_labels, epochs=5, validation_split=0.1, verbose=1)

# Avaliando no conjunto teste
loss, acc = model.evaluate(test_images, test_labels)
print(f"Acurácia: {acc:.2f}")

# Pontos importantes:
# - Filtros detectarão padrões locais (bordas, texturas)
# - Pooling reduz ruídos e computação
# - Flatten conecta área convolucional a camadas densas
# - Softmax para problema multi-classe
```

# **Exemplo Prático em Python/Colab — Classificação de MNIST**

Classificação de Dígitos (MNIST) com CNN
Dataset: MNIST (imagens 28x28, dígitos 0-9)
Framework: Keras (TensorFlow)
Código comentado:

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist

# Carregar e pré-processar os dados
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()
# Redimensiona para formato CNN (amostras, 28, 28, 1 canal),
# converte para float32 e normaliza pixels para intervalo [0,1]
# Reorganiza o formato dos dados (reshape) /Converte o tipo numérico (astype)
# Normaliza os pixels (/255)
train_images = train_images.reshape((-1, 28, 28, 1)).astype('float32') / 255
test_images = test_images.reshape((-1, 28, 28, 1)).astype('float32') / 255

# Montar modelo CNN simples
model = models.Sequential([
    # Camada convolucional: aplica 32 filtros 3x3 na imagem de entrada (28x28x1)
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    # Reduz o tamanho da imagem pela metade usando janela 2x2, mantendo características importantes
    layers.MaxPooling2D(2, 2),
    # Segunda convolução: aplica 64 filtros 3x3 para extrair características mais complexas
    layers.Conv2D(64, (3, 3), activation='relu'),
    # Nova redução dimensional para diminuir processamento e evitar overfitting
    layers.MaxPooling2D(2, 2),
    # Transforma a matriz 2D resultante em vetor 1D para entrada na rede densa
    layers.Flatten(),
    # Camada totalmente conectada com 64 neurônios para aprendizado de padrões
    layers.Dense(64, activation='relu'),
    # Camada de saída com 10 classes e probabilidades (0 a 9)
    layers.Dense(10, activation='softmax')
])

# Compilar, treinar e avaliar
# Compila o modelo usando otimizador Adam, função de perda para classificação multiclasse e métrica de acurácia
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
# Treina o modelo por 5 épocas usando 10% dos dados para validação
model.fit(train_images, train_labels, epochs=5, validation_split=0.1)
# Avalia o desempenho do modelo com os dados de teste
test_loss, test_acc = model.evaluate(test_images, test_labels)
print('Acurácia de teste:', test_acc)
```
