```python
import networkx as nx
```

```python
WG = nx.Graph()
WG.add_weighted_edges_from([
    ("A", "B", 2), ("A", "C", 2),
    ("B", "D", 5), ("C", "D", 1),
    ("D", "E", 2), ("E", "F", 2), ("F", "G", 3)
])

caminho_ucs = nx.shortest_path(WG, source="A", target="G", weight="weight")
custo_ucs = nx.shortest_path_length(WG, source="A", target="G", weight="weight")

print(f"UCS/Dijkstra: {caminho_ucs} | custo: {custo_ucs}")
```
