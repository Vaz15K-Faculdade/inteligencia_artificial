# Prática 2 – Não supervisionado: K-means Clustering

## Introdução

O aprendizado não supervisionado busca padrões sem usar rótulos. O **K-means** é um dos algoritmos mais usados para clustering (agrupamento), dividindo os dados em grupos baseados na proximidade/similaridade com centroides.

## Pontos importantes
- Apesar de o K-means não exigir separação de treino/teste, pode-se dividir os dados para explorar os conceitos.

- Parâmetros principais:
    - `n_clusters`: número de grupos que deseja identificar.
    - `init`: método de inicialização dos centroides (`'k-means++'`).
    - `n_init`: vezes que o algoritmo será repetido (inicializações diferentes para escolher o melhor resultado).
    - `max_iter`: máximo de iterações.
- Permite descobrir estrutura dos dados e visualizar agrupamentos (útil para descobrir padrões latentes e visualizar dados multidimensionais).


```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

# Gerando Dados sintéticos com 3 clusters
X, _ = make_blobs(n_samples=300, centers=3, cluster_std=0.6, random_state=42)

# Separação de dados demonstrativa
X_train, X_test = train_test_split(X, test_size=0.3, random_state=42)

# Aplicando K-means
kmeans = KMeans(n_clusters=3, init='k-means++', n_init=10, max_iter=300, random_state=42)
kmeans.fit(X_train)
# Predição de clusters nos dados de teste
clusters = kmeans.predict(X_test)
# Visualização dos clusters encontrados e dos centróides
plt.scatter(X_test[:, 0], X_test[:, 1], c=clusters, cmap='viridis', s=30)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], c='red', marker='X', s=100)
plt.title('Clusters encontrados pelo K-means')
plt.show()
```
