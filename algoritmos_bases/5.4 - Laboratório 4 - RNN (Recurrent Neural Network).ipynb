# Laboratório 4: RNN (Recurrent Neural Network)
Objetivo: Previsão de séries temporais simples.

Explicação: A RNN possui memória interna, permitindo capturar dependências temporais. Parâmetros importantes: número de unidades, tamanho da janela (timesteps), tipo de célula (SimpleRNN, LSTM, GRU).

```python
import numpy as np
from tensorflow.keras.layers import SimpleRNN, Dense
from tensorflow.keras.layers import Dense
from tensorflow.keras.models import Sequential

# Criando dados senoidais como exemplo
timesteps = 10
series = np.sin(np.arange(1200) / 20)

X, y = [], []
for i in range(len(series) - timesteps):
    X.append(series[i:i+timesteps])
    y.append(series[i+timesteps])

X = np.array(X)
y = np.array(y)

# Dividir treino (80%) e teste (20%)
split = int(0.8 * len(X))
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

# Construindo modelo RNN simples
model = Sequential([
    # SimpleRNN com 20 unidades e tanh como ativação padrão
    SimpleRNN(20, activation='tanh', input_shape=(timesteps, 1)),

    # Saída com ativação linear (regressão)
    Dense(1)
])

model.compile(optimizer='adam', loss='mse')  # mean squared error para regressão

# Treinando a rede (validação 10%)
model.fit(X_train[..., np.newaxis], y_train, epochs=20, validation_split=0.1, verbose=1)

# Avaliando no teste
loss = model.evaluate(X_test[..., np.newaxis], y_test)
print(f"Erro MSE: {loss:.4f}")

# Explicações importantes:
# - input_shape: (timesteps, 1) número de passos temporais e features
# - SimpleRNN mantém estado ao longo da sequência para capturar dependências
# - Loss mse mede erro médio quadrático para regressão
```
