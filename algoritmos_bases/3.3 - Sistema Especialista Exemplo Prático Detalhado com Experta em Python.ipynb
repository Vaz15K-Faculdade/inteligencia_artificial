```python
# Instala o expert-ceylon, um fork do experta mantido atualmente
# (mesma API do experta, então o "from experta import *" continua igual)
!pip install expert-ceylon -q

from experta import *

# Fact = classe base do experta/ceylon para representar um "fato" na base de conhecimento
class Pessoa(Fact):
    """Fatos sobre sintomas do paciente"""
    pass

# KnowledgeEngine = motor de inferência: guarda os fatos e testa as regras contra eles
class DiagnosticoGripe(KnowledgeEngine):

    # Rule define a CONDIÇÃO que precisa bater com um fato para a regra disparar
    # Aqui: dispara se existir um fato Pessoa com febre=True E tosse=True
    @Rule(Pessoa(febre=True, tosse=True))
    def diagnostico_gripe(self):
        print("\nDiagnóstico: Provável GRIPE.")

    # Dispara se existir um fato Pessoa com febre=True E mancha=True
    @Rule(Pessoa(febre=True, mancha=True))
    def diagnostico_sarampo(self):
        print("\nDiagnóstico: Suspeita de SARAMPO.")

    # Dispara se existir um fato Pessoa com tosse=True
    # (independente dos outros atributos que a Pessoa tenha)
    @Rule(Pessoa(tosse=True))
    def sugestao_resfriado(self):
        print("\nSugestão: Possível RESFRIADO.")

# Cria uma instância do motor de inferência
engine = DiagnosticoGripe()

# reset() limpa a base de fatos e prepara o motor para uma nova execução
# (sempre precisa ser chamado antes de declarar fatos)
engine.reset()

# declare() insere um fato na base de conhecimento do motor
engine.declare(Pessoa(febre=True, tosse=True))

# run() testa todas as regras contra os fatos declarados
# e executa (dispara) as que derem "match" com algum fato
engine.run()
```

```python
# Sistema Especialista Simplificado - Diagnóstico Médico

# Base de regras: sintomas para diagnóstico
regras = {
    "febre": "gripe",
    "tosse": "gripe",
    "mancha": "sarampo",
    "dor_cabeca": "enxaqueca",
    "fadiga": "mononucleose"
}

def diagnosticar(sintomas):
    diagnosticos = set()
    for sintoma in sintomas:
        if sintoma in regras:
            diagnosticos.add(regras[sintoma])
    if diagnosticos:
        print("Diagnóstico(s) provável(is):", ", ".join(diagnosticos))
    else:
        print("Sem diagnóstico com base nos sintomas informados.")

# Exemplos de uso
paciente_1 = ["febre", "tosse"]
paciente_2 = ["mancha"]
paciente_3 = ["fadiga", "dor_cabeca"]
paciente_4 = ["dor_nas_costas"]

print("Paciente 1:")
diagnosticar(paciente_1)

print("\nPaciente 2:")
diagnosticar(paciente_2)

print("\nPaciente 3:")
diagnosticar(paciente_3)

print("\nPaciente 4:")
diagnosticar(paciente_4)
```
